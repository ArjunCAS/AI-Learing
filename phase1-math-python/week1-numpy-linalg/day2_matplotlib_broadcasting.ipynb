```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1,2,3,4,5])
y = x ** 2

plt.plot(x,y)
plt.title("y is square of x")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
x = np.linspace(0,5,100)
y = x ** 2

plt.plot(x,y,color='blue',linewidth=2)
plt.title("y is square of x")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
x = np.linspace(-3, 3, 100)

plt.plot(x, x**2, label="x²", color="blue")
plt.plot(x, x**3, label="x³", color="red")
plt.plot(x, abs(x), label="|x|", color="green")

plt.title("Common ML functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
